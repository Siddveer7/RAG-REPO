```python
### Example- Semantic Search - Retrive the Similiar sentence

documents =[
    "Langchain is a framework for developing applications powered by language models",
    "It provides a set of tools and abstractions for working with LLMs",
    "Langchain enables developers to create more complex and capable applications"
]

]

query="what is langchain?"

```

```python
def semantic_search(query,documents,embeddings_models,top_k=3):
    """simple sematic search implementation"""

    ## Embed query and document

    query_embeddings=embeddings_models.embed_query(query)
    document_embeddings=embeddings_models.embed_documents(documents)


    ## Calculate the Similarity Scores
    Similarity =[]

    for i,doc_embedding in enumerate(document_embeddings):
        similarity = cosine_similarity(query_embeddings,doc_embedding)
        Similarity.append((similarity,documents[i]))

    ## Sort by Similarity

    Similarity.sort(reverse=True)
    return Similarity[:top_k]   
```

```python
query = "what is langchain?"
results = semantic_search(query, documents, embeddings)
results
```
