### What are embeddings?.

###Embeddings are nothing but representation of words into the higher dimensional numbers where the similarity of the words can be captured in the vector space.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#simplified example of 2D embeddings 

word_embeddings ={
    "cat": [0.1, 0.3],
    "kitten": [0.2, 0.4],
    "dog": [0.7, 0.8],
    "puppy": [0.6, 0.9],
    "car": [0.9, 0.1],
    "truck": [0.8, 0.2]
}
```

```python
fig, ax =plt.subplots()
for word, vec in word_embeddings.items():
    ax.scatter(vec[0], vec[1])
    ax.annotate(word, (vec[0], vec[1]))
plt.show()

ax.set_xlabel("Dimension 1")
ax.set_ylabel("Dimension 2")        
ax.set_title("2D Word Embeddings Visualization")
ax.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```

### Cosine Similarity 

### Measuring Similarity 

```python
def cosine_similarity(vec1, vec2):
    dot_product = np.dot(vec1, vec2)
    norm_vec1 = np.linalg.norm(vec1)
    norm_vec2 = np.linalg.norm(vec2)
    return dot_product / (norm_vec1 * norm_vec2)
```

```python
cosine_similarity(word_embeddings["cat"], word_embeddings["kitten"])
```
