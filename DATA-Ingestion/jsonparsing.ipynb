```python
import json 
import os
os.makedirs("data/json_files",exist_ok=True)
```

```python
json_data = {
  "product": "Laptop",
  "price": 799.99,
  "inStock": True,
  "specifications": {
    "brand": "TechBrand",
    "model": "X123",
    "features": ["Touchscreen", "8GB RAM", "256GB SSD"]
  },
  "reviews": [
    { "user": "Alice", "rating": 5, "comment": "Excellent product!" },
    { "user": "Bob", "rating": 4, "comment": "Good value for money." }
  ]
}
```

```python
json_data
```

```python
with open('data/json_files/product.json','w') as f:
    json.dump(json_data,f,indent=4)
```

```python
events = [
    {
        "event": "Product Viewed",
        "timestamp": "2023-10-01T12:00:00Z",
        "user": "Alice"
    },
    {
        "event": "Product Added to Cart",
        "timestamp": "2023-10-01T12:05:00Z",
        "user": "Bob"
    },
    {
        "event": "Checkout Started",
        "timestamp": "2023-10-01T12:10:00Z",
        "user": "Alice"
    }
]
with open('data/json_files/events.json','w') as f:
    for item in json_data:
        f.write(json.dumps(item)+"\n")
```

## Json processing Strategies

1. **Loading JSON Data**: Use `json.load()` to read JSON data from a file.
2. **Dumping JSON Data**: Use `json.dump()` to write JSON data to a file.
3. **Pretty Printing**: Use the `indent` parameter in `json.dump()` for better readability.
4. **Handling Exceptions**: Use try-except blocks to handle potential errors during file operations.
5. **Data Validation**: Validate JSON data structure before processing.

```python
from langchain_community.document_loaders import JSONLoader
import json

## Method1 : JsonLoader with JQschemea 
employee_loader = JSONLoader(
    file_path='data/json_files/product.json',
    jq_schema='.reviews[]',
    text_content=False
)
employee_docs = employee_loader.load()
print(f"Loaded {len(employee_docs)} review documents")
print("First review:", employee_docs[0].page_content if employee_docs else "No documents loaded")
print(employee_docs)
```
