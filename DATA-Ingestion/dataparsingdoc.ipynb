### word document processing 

```python
from docx import Document as DocxDocument
import os
```

```python
from langchain_community.document_loaders import Docx2txtLoader,UnstructuredWordDocumentLoader
```

```python
print(" 1 using Docx2txtLoader")
try:
    docs_loader=Docx2txtLoader("data/text_files/Siddveer_AI_Engineering_Resume.docx")
    docs = docs_loader.load()
    print(f"Loaded {len(docs)} documents(s)")
    print(f"content preview:{docs[0].page_content[:200]}...")
    print(f"metadata preview:{docs[0].metadata}")
except Exception as e:
    print("Error loading documents:", e)

```

```python
### Method2

print("\n Using UnstructuredWordDocumentLoader")
try:
    understructured_loader=UnstructuredWordDocumentLoader("data/text_files/Siddveer_AI_Engineering_Resume.docx",mode="elements")
    Unstructured_docs=understructured_loader.load()
    for i, element in enumerate(Unstructured_docs):
        print(f"\nElement {i+1}:")
        print(f"Content preview: {element.page_content[:200]}...")
        print(f"Metadata preview: {element.metadata}")
        print(f"Element {i+1} preview:")
        print(f"Content: {element.page_content[:200]}")
        print(f"Metadata: {element.metadata}")
        print(f"Element {i+1} full content:")
        print(element.page_content)
    print("Unstructured documents loaded successfully.")
except Exception as e:
    print("Error loading unstructured documents:", e)
```
